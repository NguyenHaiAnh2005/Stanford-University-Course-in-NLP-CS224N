```python
# oop in python, classes

class Animal(object):
    def __init__(self, name, age):
        self.name = name
        self.age = age
    def display(self):
        print(f"Name: {self.name} \n Age: {self.age}")
    def display_name(self):
        print(f"Name: {self.name}")
    def display_age(self):
        print(f"Age: {self.age}")
a = Animal('Dog', 20)
a.display()
```
